## Paso 4: Análisis avanzado de indicadores de negocio

### Uso de tabla temporal WITH con indicadores clave y consulta de top 10 clientes según total gastado

```python
query="""
    WITH ResumenClientes AS (
        SELECT "Customer ID",
            COUNT(Invoice) AS num_compras,
            SUM(Quantity * Price) AS total_gastado,
            MAX(InvoiceDate) AS fecha_ultima_compra
        FROM ventas
        GROUP BY "Customer ID"
    )
    SELECT * FROM ResumenClientes
    WHERE total_gastado>1000
    ORDER BY total_gastado DESC
    LIMIT 10;
    """
top_clientes=pd.read_sql(query,conn)
print(top_clientes)
```

## Top de Paises (SQL con CTE)

```python
query_paises="""
            WITH TotalGlobal AS (
                SELECT SUM(Quantity * Price) as ventas_mundiales
                FROM ventas),
            VentasPais AS (
                SELECT
                    Country, SUM(Quantity * Price) AS ingresos_pais
                FROM ventas
                GROUP BY Country)
            SELECT 
                vp.Country,
                vp.ingresos_pais,
                (vp.ingresos_pais*100/tg.ventas_mundiales) as porcentaje_del_total
                FROM VentasPais vp, TotalGlobal tg
                ORDER BY vp.ingresos_pais DESC
                LIMIT 10;
                """
df_paises=pd.read_sql_query(query_paises,conn)
print(df_paises)
```

```python
plt.figure(figsize=(12,6))
sns.set_style("whitegrid")

ax=sns.barplot(x='ingresos_pais',y='Country',data=df_paises,palette='viridis')
for p in ax.patches:
    width=p.get_width()
    plt.text(p.get_width(),p.get_y()+p.get_height()/2.0,
        f' ${width:,.0f}',va='center')

plt.title('Top 10 paises por generación de ingresos', fontsize=15)
plt.xlabel('Ingresos Totales (USD)',fontsize=12)
plt.ylabel('País', fontsize=12)
plt.show()
```

#### Podemos identificar que el principal mercado con casi un 83% de las ventas corresponde a United Kingdom. Centraré el resto de mi análisis en este país

### Los 10 productos más vendidos en UK usando Window Functions (nuestro principal mercado)

```python
query_uk_top="""
            WITH VentasProductos AS (
            SELECT
                Description,
                SUM(Quantity * Price) as total_ventas
            FROM ventas
            WHERE Country ='United Kingdom'
            GROUP BY Description
            )
            SELECT
                Description,
                total_ventas,
                RANK() OVER(ORDER BY total_ventas DESC) as ranking
            FROM VentasProductos
            LIMIT 10;
            """
df_uk=pd.read_sql_query(query_uk_top,conn)
print(df_uk)
            
```

```python
df_uk.set_index('Description')['total_ventas'].plot(kind='barh',color='skyblue')
plt.title('Ranking productos más vendidos en United Kingdom')
plt.xlabel('Total vendido en USD')
plt.ylabel('Producto')
plt.show()
```

### Análisis de ventas mensuales en el mercado de UK para detectar tendencias y patrones

```python
query_mensual="""
            SELECT
            strftime('%Y-%m',InvoiceDate) as mes,
            SUM(Quantity * Price) as ventas_mensuales
            FROM ventas
            WHERE Country='United Kingdom'
            GROUP BY mes
            ORDER BY mes;
            """
df_mensual=pd.read_sql_query(query_mensual,conn)
plt.figure(figsize=(10,5))
plt.plot(df_mensual['mes'], df_mensual['ventas_mensuales'],
         marker='o',linestyle='-')
plt.xticks(rotation=45)
plt.title('Evolución de Ventas Mensuales en UK')
plt.grid(True,linestyle='--',
        alpha=0.6)
plt.show()
```

#### Se observa un patrón ciclo con picos de ventas entre los meses de Septiembre a Noviembre, para decaer en septiembre y alcanzar su mínimo anual entre Enero y Febrero. Entre Marzo y Agosto se observa una variabilidad normal.

## Análisis Churn Rate (Fuga de Clientes)

```python
#Para este caso defino un cliente en fuga (Churn) como alguien que realizó compras en Noviembre, pero no registró ninguna compra en Enero
#Para este efecto hago una comparación de dos listas con los ID de cliente en noviembre y enero

query_churn="""
            WITH Clientes_Noviembre AS (
            SELECT DISTINCT "Customer ID"
            FROM ventas
            WHERE strftime ('%Y-%m', InvoiceDate) = '2010-11'),
            Clientes_Enero AS (
            SELECT DISTINCT "Customer ID"
            FROM ventas
            WHERE strftime('%Y-%m',InvoiceDate)='2011-01')
            SELECT COUNT(*) AS clientes_que_no_volvieron
            FROM Clientes_Noviembre
            WHERE "Customer ID" NOT IN (SELECT "Customer ID" FROM Clientes_Enero);
            """
resultado_churn=pd.read_sql_query(query_churn,conn)
print(f"Clientes que compraron en Noviembre pero NO en Enero: {resultado_churn.iloc[0,0]}")
```

#### Si bien el Churn es un indicador muy poderoso, en este dataset de retail no es el más representativo ya que los clientes más que fugarse, tienen patrones de compra más aleatorios que no necesariamente son indicativos de que dicho cliente compró a la competencia. Es por esto que hago uso de un indicador complementario

### Implementación de indicador RFM para comprender el comportamiento de los clientes

```python
query_base_rfm="""
                SELECT
                    "Customer ID",
                    MAX(InvoiceDate) AS ultima_compra,
                    COUNT(DISTINCT Invoice) AS frecuencia,
                    SUM(Quantity * Price) AS monto_total
                FROM ventas
                WHERE Country = 'United Kingdom'
                GROUP BY "Customer ID";
                """
df_rfm=pd.read_sql_query(query_base_rfm,conn)
df_rfm['ultima_compra']=pd.to_datetime(df_rfm['ultima_compra'])
```

```python
# defino la fecha de referencia de acuerdo con la disponibilidad en el dataset
fecha_referencia=df_rfm['ultima_compra'].max()+pd.Timedelta(days=1)

#calculo la recencia: días transcurridos desde la última compra
df_rfm['recencia']=(fecha_referencia - df_rfm['ultima_compra']).dt.days

print(df_rfm[['Customer ID','ultima_compra','recencia']].head())
```

```python
#Una visualización para conocer la distribución de estos datos
plt.figure(figsize=(10,5))
plt.hist(df_rfm['recencia'],bins=30,color='orange',edgecolor='black')
plt.title('Distribución de la Recencia')
plt.xlabel('Días')
plt.ylabel('Cantidad de Clientes')
plt.show()
```

```python
#usaré puntajes de 1 al 5 para los tres indicadores que estoy considerando (recencia, frecuencia y monto)
#Para la Recencia el valor más bajo es mejor, es decir el cliente mejor calificado tendrá un 5
df_rfm['R']=pd.qcut(df_rfm['recencia'],5,labels=[5,4,3,2,1])
df_rfm['F']=pd.qcut(df_rfm['frecuencia'].rank(method='first'),5,labels=[1,2,3,4,5])
df_rfm['M']=pd.qcut(df_rfm['monto_total'],5,labels=[1,2,3,4,5])

#Creo puntajes combinados como texto
df_rfm['RFM_Score']= df_rfm['R'].astype(str) + df_rfm['F'].astype(str) + df_rfm['M'].astype(str)
print(df_rfm[['Customer ID','RFM_Score']].head())
```

```python
#Aquí defino una función para clasificar a los clientes según la metodología RFM, considerar que el valor de M para esta clasificación no será considerado

def segmentar_rfm(df):
    r=int(df['R'])
    f=int(df['F'])

    if r>=4 and f>=4:
        return 'Campeones'
    elif r>=3 and f>=3:
        return 'Leales'
    elif r>=4 and f <3:
        return 'Nuevos / Promesas'
    elif r< 3 and f>=3:
        return 'En Riesgo / No perder'
    else: 
        return 'Perdidos / Inactivos'

# Aplico la función a cada fila
df_rfm['Segmento']=df_rfm.apply(segmentar_rfm,axis=1)
#conteo de clientes por grupo
resumen_segmentos=df_rfm['Segmento'].value_counts().reset_index()
resumen_segmentos.columns=['Segmento','Cantidad_Clientes']

total_clientes=resumen_segmentos['Cantidad_Clientes'].sum()
resumen_segmentos['Porcentaje']=(resumen_segmentos['Cantidad_Clientes']/total_clientes)*100

resumen_segmentos['Porcentaje']=resumen_segmentos['Porcentaje'].map('{:.2f}%'.format)
print(resumen_segmentos)
#visualización de lo anterior
plt.figure(figsize=(10,6))
resumen.plot(kind='barh',color='teal')

plt.title('Distribución de Segmentos de Clientes', fontsize=14)
plt.xlabel('Número de Clientes')
plt.ylabel('Segmento')
plt.grid(axis='x',linestyle='--',alpha=0.7)
plt.tight_layout()
plt.show()
```

### El indicador RFM (Recencia, Frecuencia y Monto) generó una clasificación clara de los clientes en United Kingdom de acuerdo a sus comportamiento. Esta información es de gran ayuda para estratégias de Marketing focalizado
